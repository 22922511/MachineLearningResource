```python
from sklearn.feature_selection import SelectKBest, f_regression
from sklearn.linear_model import LinearRegression, Lasso, Ridge
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures as pb
from xgboost import XGBRegressor
from sklearn.metrics import mean_absolute_percentage_error as mape
import pandas as pd

deg = 1# degree for polynomial features

# Define models
model1 = make_pipeline(pb(deg), LinearRegression())
model2 = make_pipeline(pb(deg), Lasso(alpha=0.5))
model3 = make_pipeline(pb(deg), Ridge(alpha=0.5))
model4 = XGBRegressor()

def pre(x_train, y_train, x_test, y_test):
    # Convert y to 1D if it's a DataFrame
    if isinstance(y_train, pd.DataFrame):
        y_train = y_train.values.ravel()
        y_test = y_test.values.ravel()

    # SelectKBest for feature selection
    selector = SelectKBest(score_func=f_regression, k=20)
    selector.fit(x_train, y_train)

    selected_columns = x_train.columns[selector.get_support()]
    print("Selected Features:", selected_columns.tolist())

    x_train_selected = selector.transform(x_train)
    x_test_selected = selector.transform(x_test)

    print(f"x_train: {x_train_selected.shape}, y_train: {y_train.shape}, x_test: {x_test_selected.shape}, y_test: {y_test.shape}")

    # Fit models
    model1.fit(x_train_selected, y_train)
    model2.fit(x_train_selected, y_train)
    model3.fit(x_train_selected, y_train)
    model4.fit(x_train_selected, y_train)

    # Predict
    y_pred_test1 = model1.predict(x_test_selected)
    y_pred_train1 = model1.predict(x_train_selected)

    y_pred_test2 = model2.predict(x_test_selected)
    y_pred_train2 = model2.predict(x_train_selected)

    y_pred_test3 = model3.predict(x_test_selected)
    y_pred_train3 = model3.predict(x_train_selected)

    y_pred_test4 = model4.predict(x_test_selected)
    y_pred_train4 = model4.predict(x_train_selected)

    # Print MAPE results
    print("LinearReg     :", mape(y_test, y_pred_test1), mape(y_train, y_pred_train1))
    print("Lasso         :", mape(y_test, y_pred_test2), mape(y_train, y_pred_train2))
    print("Ridge         :", mape(y_test, y_pred_test3), mape(y_train, y_pred_train3))
    print("XGBoost       :", mape(y_test, y_pred_test4), mape(y_train, y_pred_train4))
```
